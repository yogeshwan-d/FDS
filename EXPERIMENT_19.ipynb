```python
# 19) Experiment to understand Logistic Regression for a given data set.
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
# Sample Social Network Ads dataset (400 records, same columns as Social_Network_Ads.csv)
np.random.seed(7)
head_rows = [[15624510, 'Male', 19, 19000, 0], [15810944, 'Male', 35, 20000, 0], [15668575, 'Female', 26, 43000, 0],
             [15603246, 'Female', 27, 57000, 0], [15804002, 'Male', 19, 76000, 0], [15728773, 'Male', 27, 58000, 0],
             [15598044, 'Female', 27, 84000, 0], [15694829, 'Female', 32, 150000, 1], [15600575, 'Male', 25, 33000, 0],
             [15727311, 'Female', 35, 65000, 0], [15570769, 'Female', 26, 80000, 0], [15606274, 'Female', 26, 52000, 0],
             [15746139, 'Male', 20, 86000, 0], [15704987, 'Male', 32, 18000, 0], [15628972, 'Male', 18, 82000, 0],
             [15697686, 'Male', 29, 80000, 0], [15733883, 'Male', 47, 25000, 1], [15617482, 'Male', 45, 26000, 1],
             [15704583, 'Male', 46, 28000, 1], [15621083, 'Female', 48, 29000, 1]]
tail_rows = [[15691863, 'Female', 46, 41000, 1], [15706071, 'Male', 51, 23000, 1], [15654296, 'Female', 50, 20000, 1],
             [15755018, 'Male', 36, 33000, 0], [15594041, 'Female', 49, 36000, 1]]
age = np.random.randint(18, 61, 375)
salary = np.random.randint(15, 151, 375) * 1000
prob = 1 / (1 + np.exp(-(0.25 * (age - 42) + (salary - 90000) / 15000)))
middle_rows = [[int(np.random.randint(15566689, 15815237)), str(np.random.choice(['Male', 'Female'])), int(age[i]), int(salary[i]),
                int(np.random.rand() < prob[i])] for i in range(375)]
df = pd.DataFrame(head_rows + middle_rows + tail_rows, columns=['User ID', 'Gender', 'Age', 'EstimatedSalary', 'Purchased'])
print(df)
print(df.head())
# Separate the features (Age, EstimatedSalary) and the label (Purchased)
features = df.iloc[:, [2, 3]].values
label = df.iloc[:, 4].values
print(features)
print(label)
# Find the random states where the test score is greater than the train score
for i in range(1, 401):
    x_train, x_test, y_train, y_test = train_test_split(features, label, test_size=0.2, random_state=i)
    model = LogisticRegression()
    model.fit(x_train, y_train)
    train_score = model.score(x_train, y_train)
    test_score = model.score(x_test, y_test)
    if test_score > train_score:
        print("Test {} Train{} Random State {}".format(test_score, train_score, i))
# Train the final Logistic Regression model
x_train, x_test, y_train, y_test = train_test_split(features, label, test_size=0.2, random_state=42)
finalModel = LogisticRegression()
finalModel.fit(x_train, y_train)
# Training and testing scores
print(finalModel.score(x_train, y_train))
print(finalModel.score(x_test, y_test))
# Classification report
print(classification_report(label, finalModel.predict(features)))
```
