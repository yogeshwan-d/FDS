```python
# 10) Experiment to understand array function in Data science.
import numpy as np
# Randomly generate 9 numbers between 1 to 100
array = np.random.randint(1, 100, 9)
print(array)
# Square root of each element
print(np.sqrt(array))
# Number of dimensions
print(array.ndim)
# Reshape the 1D array into a 3x3 array
new_array = array.reshape(3, 3)
print(new_array)
print(new_array.ndim)
# Flatten the 2D array back to 1D
print(new_array.ravel())
newm = new_array.reshape(3, 3)
print(newm)
# Slicing the array
print(newm[2, 1:3])
print(newm[1:2, 1:3])
print(new_array[0:3, 0:0])
print(new_array[0:2, 0:1])
print(new_array[0:3, 0:1])
print(new_array[1:3])
```
