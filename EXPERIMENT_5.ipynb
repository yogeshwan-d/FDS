```python
# 5) Data Collection and Initial Exploration - To collect, load, and perform initial exploration of the diabetes dataset.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# 1. Data Collection:
# Sample diabetes dataset (768 records, same columns as diabetes.csv)
np.random.seed(42)
n = 768
# Load the dataset
db = pd.DataFrame({
'Pregnancies': np.clip(np.random.poisson(3.8, n), 0, 17),
'Glucose': np.clip(np.random.normal(121, 32, n), 0, 199).astype(int),
'BloodPressure': np.clip(np.random.normal(69, 19, n), 0, 122).astype(int),
'SkinThickness': np.clip(np.random.normal(20.5, 16, n), 0, 99).astype(int),
'Insulin': np.clip(np.random.exponential(80, n), 0, 846).astype(int),
'BMI': np.round(np.clip(np.random.normal(32, 7.9, n), 0, 67.1), 1),
'DiabetesPedigreeFunction': np.round(np.clip(np.random.exponential(0.47, n), 0.078, 2.42), 3),
'Age': np.clip(np.random.exponential(12, n) + 21, 21, 81).astype(int),
'Outcome': np.random.binomial(1, 0.35, n)})
# Display the first few rows
print(db.head())
# 2. Initial Exploration:
# Dataset structure and summary statistics
print(db.info())
print(db.describe())
# Visualize the distribution of each feature
db.hist(bins=50, figsize=(20, 15))
plt.show()
# Pairplot to visualize relationships
sns.pairplot(db)
plt.show()
```
