```python
# 18) Experiment to understand KNN algorithm for a given data set.
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
# Load the Iris dataset (same as Iris.csv)
iris = load_iris()
df = pd.DataFrame(iris.data, columns=['sepal.length', 'sepal.width', 'petal.length', 'petal.width'])
df['variety'] = np.array(['Setosa', 'Versicolor', 'Virginica'])[iris.target]
df.info()
print(df.variety.value_counts())
print(df.head())
# Separate the features and the label
features = df.iloc[:, :-1].values
label = df.iloc[:, 4].values
# Split the data into training and testing sets
xtrain, xtest, ytrain, ytest = train_test_split(features, label, test_size=.2, random_state=1)
# Train the KNN model with 5 neighbors
model_KNN = KNeighborsClassifier(n_neighbors=5)
model_KNN.fit(xtrain, ytrain)
# Training and testing scores
print(model_KNN.score(xtrain, ytrain))
print(model_KNN.score(xtest, ytest))
# Confusion matrix
print(confusion_matrix(label, model_KNN.predict(features)))
# Classification report
print(classification_report(label, model_KNN.predict(features)))
```
